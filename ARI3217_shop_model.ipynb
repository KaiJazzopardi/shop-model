This notebook trains a small model that predicts whether a product will sell, given its price.
It works, but it is written the way many exploratory notebooks are: everything happens in
cells, the data is changed in place, and the settings are hardcoded. **That is deliberate.**
Over this session you will turn it into a proper Python package.

The data (`prices.csv`) holds 500 synthetic product listings from a small homeware and gift shop:
a short description of the item, its price before VAT, in euro, and whether it sold (1) or not (0).
The model uses the price only. Malta's standard VAT rate of 18% is added before training.

Follow the steps in the cell at the bottom of the notebook *after* running it once from the top.

1) Loading the pre-requisite packages

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

2. Load the dataset

```python
df = pd.read_csv("https://raw.githubusercontent.com/ARI3217/datasets/main/prices.csv")
display (df.head())
```

3. Adding the VAT to the prices (overwriting the original data)

```python
df["price"] = df["price"] * 1.18
display (df.head ())
```

4. Train the model using Logistic Regression

```python
# Cell 3: train the model
model = LogisticRegression(random_state=42)
model.fit(df[["price"]], df["sold"]);
print ("Model Trained")
```

4. Predict the sales for a new item - Ceramic fruit bowl priced at €20.00 exc. VAT

```python
new_item = pd.DataFrame({
    "description": ["Ceramic fruit bowl, hand-painted"],
    "price": [20.00 * 1.18],
})
probability = model.predict_proba(new_item[["price"]])[0][1]

print(f"Average training price: EUR {df['price'].mean():.2f}")
print(f"New item: {new_item['description'][0]}")
print(f"Probability that the new item sells: {probability:.3f}")
```
